## 1. Load Kaggle disease-symptom dataset


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/tmp/dataset/dataset.csv')
print(f'Shape: {df.shape}')
print(f'Dtypes:\n{df.dtypes}')
print(f'Sample rows:\n{df.head()}')
```

## 2. Disease class distribution bar chart


```python
plt.figure(figsize=(12, 8))
df['Disease'].value_counts().sort_values().plot(kind='barh')
plt.title('Disease Class Distribution')
plt.tight_layout()
plt.show()
```

## 4. Top 30 symptom frequency bar chart


```python
symptom_cols = [c for c in df.columns if 'Symptom' in c]
symptoms = df[symptom_cols].values.flatten()
symptoms = pd.Series(symptoms).dropna().str.strip().str.replace(' ', '_')

plt.figure(figsize=(12, 8))
symptoms.value_counts().head(30).sort_values().plot(kind='barh')
plt.title('Top 30 Symptom Frequency')
plt.tight_layout()
plt.show()
```

## 5. Symptom co-occurrence heatmap


```python
top_20 = symptoms.value_counts().head(20).index
symptoms_df = df[symptom_cols].apply(lambda x: pd.Series(x.dropna().unique()), axis=1)
symptoms_one_hot = pd.get_dummies(symptoms_df.stack()).groupby(level=0).max()

available_top = [s for s in top_20 if s in symptoms_one_hot.columns]
co_occ = symptoms_one_hot[available_top].T.dot(symptoms_one_hot[available_top])

plt.figure(figsize=(10, 8))
sns.heatmap(co_occ, cmap='viridis')
plt.title('Symptom Co-occurrence (Top 20)')
plt.tight_layout()
plt.show()
```

## 6. Unique symptoms per disease
